# NB01 - Bronze Data Ingestion

## Objective

This notebook ingests the generated Parquet files from the **Files/bronze** folder into the Bronze Lakehouse as Delta tables.

### Source

- LH_Bronze/Files/bronze

### Destination

- LH_Bronze/Tables

At the end of this notebook, every Parquet file will be available as a managed Delta table in the Bronze layer.

```python
# ==========================================================
# Import Libraries
# ==========================================================

from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
```

## Bronze File Location

This section defines the location of the uploaded Parquet files.

All files were uploaded into:

`LH_Bronze/Files/bronze`    

```python
# ==========================================================
# Bronze Folder Path
# ==========================================================

bronze_path = "Files/bronze"

print(bronze_path)
```

## Tables to Import

The following Parquet files will be converted into managed Delta tables inside the Bronze Lakehouse.

```python
# ==========================================================
# List of Tables
# ==========================================================

tables = [

    "DimRegion",

    "DimStore",

    "DimCustomer",

    "DimSupplier",

    "DimProduct",

    "DimEmployee",

    "DimPromotion",

    "DimDate",

    "FactSales",

    "FactInventory",

    "FactReturns"

]

print(f"Total Tables : {len(tables)}")
```

## Ingest Parquet Files into Bronze Tables

### Objective

This step performs the following actions for every dataset:

- Reads the Parquet file from **Files/bronze**
- Creates or replaces a managed Delta table in **LH_Bronze**
- Registers the table in the Lakehouse

After execution, the tables will appear under:

**LH_Bronze → Tables**

```python
# ==========================================================
# Ingest Bronze Files
# ==========================================================

for table in tables:

    print(f"\nLoading {table}...")

    df = spark.read.parquet(f"{bronze_path}/{table}.parquet")

    (
        df.write
        .mode("overwrite")
        .format("delta")
        .option("overwriteSchema", "true")
        .saveAsTable(table)
    )

    print(f"{table} loaded successfully.")
```

## Verify the Bronze Tables

### Objective

Verify that all tables have been successfully created in the Bronze Lakehouse and display the row count for each table.

```python
# ==========================================================
# Verify Tables
# ==========================================================

for table in tables:

    count = spark.table(table).count()

    print(f"{table:<20} {count:>12,}")
```
