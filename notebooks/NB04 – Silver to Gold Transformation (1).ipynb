# NB04 – Silver to Gold Transformation

## Objective

Transform the cleaned Silver layer into a business-ready Gold layer.

The Gold layer contains curated dimension tables, fact tables, and aggregated business datasets optimized for analytics and reporting.

```python
# ==========================================================
# Import Required Libraries
# ==========================================================

from pyspark.sql import functions as F
from pyspark.sql.types import *
```

## Configure Silver Layer Path

### Objective

Define the base path of the Silver Lakehouse.

All Silver tables will be read from this location before being transformed into Gold tables.

```python
# ==========================================================
# Silver Lakehouse Base Path
# ==========================================================

silver_base_path = (
    "abfss://EnterpriseRetailAnalytics@onelake.dfs.fabric.microsoft.com/"
    "LH_Silver.Lakehouse/Tables/dbo"
)

print("Silver Base Path Configured Successfully")
```

## Read Dimension Tables from Silver

### Objective

Read all curated dimension tables from the Silver Lakehouse.

These tables have already been validated and standardized in the Silver layer and will be promoted to the Gold layer for business consumption.

```python
# ==========================================================
# Read Dimension Tables from Silver
# ==========================================================

dim_date = spark.read.format("delta").load(f"{silver_base_path}/dimdate")

dim_customer = spark.read.format("delta").load(f"{silver_base_path}/dimcustomer")

dim_product = spark.read.format("delta").load(f"{silver_base_path}/dimproduct")

dim_store = spark.read.format("delta").load(f"{silver_base_path}/dimstore")

dim_region = spark.read.format("delta").load(f"{silver_base_path}/dimregion")

dim_employee = spark.read.format("delta").load(f"{silver_base_path}/dimemployee")

dim_supplier = spark.read.format("delta").load(f"{silver_base_path}/dimsupplier")

dim_promotion = spark.read.format("delta").load(f"{silver_base_path}/dimpromotion")

print("All Silver Dimension Tables Loaded Successfully")
```

## Verify Dimension Tables

### Objective

Verify that all dimension tables have been loaded successfully by checking their record counts.

```python
# ==========================================================
# Verify Dimension Tables
# ==========================================================

dimension_tables = {
    "DimDate": dim_date,
    "DimCustomer": dim_customer,
    "DimProduct": dim_product,
    "DimStore": dim_store,
    "DimRegion": dim_region,
    "DimEmployee": dim_employee,
    "DimSupplier": dim_supplier,
    "DimPromotion": dim_promotion
}

for table_name, df in dimension_tables.items():
    print(f"{table_name}: {df.count()} rows")
```

## Create Gold Dimension Tables

### Objective

Write the curated dimension tables from the Silver layer into the Gold Lakehouse.

These tables are optimized for business reporting and will serve as dimension tables in the enterprise star schema.

```python
# ==========================================================
# Write Dimension Tables to Gold Lakehouse
# ==========================================================

dimension_tables = {
    "DimDate": dim_date,
    "DimCustomer": dim_customer,
    "DimProduct": dim_product,
    "DimStore": dim_store,
    "DimRegion": dim_region,
    "DimEmployee": dim_employee,
    "DimSupplier": dim_supplier,
    "DimPromotion": dim_promotion
}

for table_name, df in dimension_tables.items():

    (
        df.write
          .mode("overwrite")
          .format("delta")
          .saveAsTable(table_name)
    )

    print(f"{table_name} written successfully.")
```

## UPDATED CODE

```python
# ==========================================================
# Write Dimension Tables to Gold Lakehouse
# ==========================================================

dimension_tables = {
    "DimDate": dim_date,
    "DimCustomer": dim_customer,
    "DimProduct": dim_product,
    "DimStore": dim_store,
    "DimRegion": dim_region,
    "DimEmployee": dim_employee,
    "DimSupplier": dim_supplier,
    "DimPromotion": dim_promotion
}

for table_name, df in dimension_tables.items():

    (
        df.write
          .mode("overwrite")
          .format("delta")
          .option("overwriteSchema", "true")
          .saveAsTable(table_name)
    )

    print(f"{table_name} written successfully.")
```

Verify that all dimension tables have been successfully created in the Gold Lakehouse.

## Read Fact Tables from Gold

### Objective

Load the Gold fact and dimension tables required to build business aggregate tables.

These tables will be used to calculate daily business KPIs. 

```python
# ==========================================================
# Read Silver Fact Tables
# ==========================================================

fact_sales = (
    spark.read
         .format("delta")
         .load(f"{silver_base_path}/factsales")
)

fact_inventory = (
    spark.read
         .format("delta")
         .load(f"{silver_base_path}/factinventory")
)

fact_returns = (
    spark.read
         .format("delta")
         .load(f"{silver_base_path}/factreturns")
)

# ==========================================================
# Read Date Dimension
# ==========================================================

dim_date = (
    spark.read
         .format("delta")
         .load(f"{silver_base_path}/dimdate")
)

print("Silver tables loaded successfully.")
```

### <mark>"Write Fact Tables to Gold Lake House" will be done at the end after creating aggregated tables</mark> 

```python
# ==========================================================
# Create Daily Sales Summary
# ==========================================================

gold_sales_daily = (

    fact_sales

    .groupBy("SalesDateKey")

    .agg(

        F.sum("SalesAmount").alias("TotalSales"),

        F.sum("NetSales").alias("TotalNetSales"),

        F.sum("DiscountAmount").alias("TotalDiscount"),

        F.sum("Quantity").alias("TotalQuantity"),

        F.count("SalesKey").alias("TotalOrders")

    )

)

print("GoldSalesDaily Created")

gold_sales_daily.show(10, truncate=False)
```

## Enrich GoldSalesDaily with Date Dimension

### Objective

Enhance the daily sales summary by joining it with the Date Dimension.

This transformation adds business-friendly calendar attributes such as:

- Date
- Year
- Quarter
- Month
- Week Number
- Fiscal Year

These attributes make the table ready for reporting without requiring additional joins in Power BI.

```python
# ==========================================================
# Join GoldSalesDaily with Date Dimension
# ==========================================================

gold_sales_daily = (

    gold_sales_daily

    .join(

        dim_date,

        gold_sales_daily.SalesDateKey == dim_date.DateKey,

        "left"

    )

    .drop(dim_date.DateKey)

)

print("GoldSalesDaily enriched with Date Dimension.")

gold_sales_daily.select(

    "SalesDateKey",
    "Date",
    "Year",
    "Quarter",
    "MonthNumber",
    "MonthName",
    "WeekNumber",
    "FiscalYear",
    "TotalSales",
    "TotalNetSales",
    "TotalDiscount",
    "TotalQuantity",
    "TotalOrders"

).show(10, truncate=False)
```

## Write GoldSalesDaily to the Gold Lakehouse

### Objective

Persist the enriched GoldSalesDaily table into the Gold Lakehouse.

This table serves as a curated, business-ready dataset for reporting and dashboarding.

```python
# ==========================================================
# Write GoldSalesDaily to Gold Lakehouse
# ==========================================================

(
    gold_sales_daily.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable("GoldSalesDaily")
)

print("GoldSalesDaily written successfully.")
```

## UPDATED CODE

```python
# ==========================================================
# Write GoldSalesDaily to Gold Lakehouse
# ==========================================================

(
    gold_sales_daily.write
    .mode("overwrite")
    .format("delta")
    .option("overwriteSchema", "true")
    .saveAsTable("GoldSalesDaily")
)

print("GoldSalesDaily written successfully.")
```

## Create GoldSalesMonthly

### Objective

Create a monthly sales summary by aggregating the daily sales data.

The table provides monthly business KPIs for trend analysis, executive reporting, and Power BI dashboards.

Metrics calculated:

- Total Sales
- Total Net Sales
- Total Discount
- Total Quantity Sold
- Total Orders

```python
# ==========================================================
# Create Monthly Sales Summary
# ==========================================================

gold_sales_monthly = (

    gold_sales_daily

    .groupBy(
        "Year",
        "Quarter",
        "MonthNumber",
        "MonthName",
        "FiscalYear"
    )

    .agg(

        F.sum("TotalSales").alias("MonthlySales"),

        F.sum("TotalNetSales").alias("MonthlyNetSales"),

        F.sum("TotalDiscount").alias("MonthlyDiscount"),

        F.sum("TotalQuantity").alias("MonthlyQuantity"),

        F.sum("TotalOrders").alias("MonthlyOrders")

    )

    .orderBy("Year", "MonthNumber")

)

print("GoldSalesMonthly Created Successfully")

gold_sales_monthly.show(20, truncate=False)
```

## Write GoldSalesMonthly

### Objective

Write the monthly sales summary into the Gold Lakehouse.

This table is optimized for executive dashboards, monthly trend analysis, and year-over-year reporting.

```python
# ==========================================================
# Write GoldSalesMonthly
# ==========================================================

(
    gold_sales_monthly.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable("GoldSalesMonthly")
)

print("GoldSalesMonthly written successfully.")
```

## Create GoldProductPerformance

### Objective

Create a product performance summary by combining sales transactions with product attributes.

This table provides business-ready KPIs for each product and includes descriptive product information.

Metrics calculated:

- Total Sales
- Total Net Sales
- Total Discount
- Total Quantity Sold
- Total Orders

```python
# ==========================================================
# Create Product Performance Summary
# ==========================================================

gold_product_performance = (

    fact_sales

    .groupBy("ProductKey")

    .agg(

        F.sum("SalesAmount").alias("TotalSales"),

        F.sum("NetSales").alias("TotalNetSales"),

        F.sum("DiscountAmount").alias("TotalDiscount"),

        F.sum("Quantity").alias("TotalQuantity"),

        F.count("SalesKey").alias("TotalOrders")

    )

)

print("GoldProductPerformance Summary Created")

gold_product_performance.show(10, truncate=False)
```

## Enrich GoldProductPerformance

### Objective

- Product Name
- Category
- Sub Category
- Brand

These attributes make the table ready for business reporting.

```python
# ==========================================================
# Read Product Dimension
# ==========================================================

dim_product = (
    spark.read
         .format("delta")
         .load(f"{silver_base_path}/dimproduct")
)

# ==========================================================
# Join Product Attributes
# ==========================================================

gold_product_performance = (

    gold_product_performance

    .join(

        dim_product,

        "ProductKey",

        "left"

    )

)

gold_product_performance.select(

    "ProductKey",
    "ProductName",
    "Category",
    "SubCategory",
    "Brand",
    "TotalSales",
    "TotalNetSales",
    "TotalQuantity",
    "TotalOrders"

).show(10, truncate=False)
```

## Create GoldInventorySummary

### Objective

Create an inventory summary by aggregating inventory transactions at the Product and Store level.

This table provides business-ready inventory KPIs.

Metrics calculated:

- Opening Stock
- Stock Received
- Stock Sold
- Closing Stock

```python
gold_inventory_summary = (

    fact_inventory

    .groupBy(
        "ProductKey",
        "StoreKey"
    )

    .agg(

        F.sum("OpeningStock").alias("OpeningStock"),
        F.sum("StockReceived").alias("StockReceived"),
        F.sum("StockSold").alias("StockSold"),
        F.sum("ClosingStock").alias("ClosingStock")

    )

)
```

## Enrich GoldInventorySummary

### Objective

Join the inventory summary with Product and Store dimensions.

This adds descriptive attributes such as:

- Product Name
- Category
- Brand
- Store Name
- Store Type
- City
- State

```python
# ==========================================================
# Join Product and Store Dimensions
# ==========================================================

gold_inventory_summary = (

    gold_inventory_summary

    .join(

        dim_product.select(
            "ProductKey",
            "ProductName",
            "Category",
            "SubCategory",
            "Brand"
        ),

        "ProductKey",
        "left"

    )

    .join(

        dim_store.select(
            "StoreKey",
            "StoreName",
            "StoreType",
            "City",
            "State"
        ),

        "StoreKey",
        "left"

    )

)

gold_inventory_summary.show(10, truncate=False)
```

## Write GoldInventorySummary

### Objective

Write the inventory summary into the Gold Lakehouse.

This table supports:

```python
# ==========================================================
# Write GoldInventorySummary
# ==========================================================

(
    gold_inventory_summary.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable("GoldInventorySummary")
)

print("GoldInventorySummary written successfully.")
```
